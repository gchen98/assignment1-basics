```python
# batch size × total step count × context length = 32 × 5000 × 256 =
# 40, 960, 000 tokens, which takes 1 hour and 22 minutes on cpu and 36 minutes on mps.
# At step 5000, we achieve a validation loss of 1.80.

import custom_modules
import custom_training
import io_utils

batch_size = 32
steps = 5
context_length = 128
embedding_dim = 64
num_layers=4
num_heads = 16
feed_forward_dim = 1344
rope_theta = 10000

infile = '../data/TinyStoriesV2-GPT4-train.npy'
vocab = io_utils.deserialize_vocab("../vocab_tinystories_sample_5M.txt")
vocab_size = len(vocab)




dataset = np.load(infile,mmap_mode = 'r')
#     def __init__(self,vocab_size:int,context_length:int,
#                  num_layers:int,d_model:int,num_heads:int,
#                  d_ff:int,rope_theta:float):
transformer_lm = custom_modules.TransformerLm(vocab_size = vocab_size,
                                              context_length = context_length,
                                              num_layers = num_layers,
                                              d_model = embedding_dim,
                                              num_heads = num_heads,
                                              d_ff = feed_forward_dim,
                                              rope_theta = rope_theta
                                              )
inputs,targets = custom_training.get_batch(dataset = dataset,batch_size = batch_size,context_length = context_length,device="cpu")
logger.debug(f"input is {inputs.shape}")
model_output = transformer_lm(inputs)
logger.debug(f"output is \n{model_output.shape} targets is \n{targets}")

cross_entropy_loss = custom_modules.CrossEntropyLoss()
loss = cross_entropy_loss(logits = model_output,targets = targets)
logger.debug(f"Loss is {loss.shape}\n{loss}")

opt = custom_training.AdamW( params = transformer_lm.parameters(),
                             lr = 1e-3,
                             weight_decay = 0.01,betas = (0.9,0.999),eps = 1e-8)


for t in range(steps):
     opt.zero_grad() # Reset the gradients for all learnable parameters
     model_output = transformer_lm(inputs)
     loss = cross_entropy_loss(logits = model_output,targets = targets)
     # loss = (weights ** 2).mean() # Compute a scalar loss value.
     logger.debug(loss.cpu().item())
     loss.backward() # Run backward pass, which computes gradients
     opt.step()

# log_softmax = custom_modules.LogSoftmax()
# log_softmax_output = log_softmax(in_features = model_output,dim=1)
# logger.debug(f"log_softmax {log_softmax_output.shape}  \n{log_softmax_output}")
#
# for step in steps:
#     x,y = custom_training.get_batch(dataset = dataset, batch_size = batch_size,
#                                     context_length=context_length,device="cpu")
#
#     print(f"batches {x.shape} {y.shape}")
```
