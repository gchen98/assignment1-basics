```python
import time

import numpy as np
from sympy import to_number_field

from cs336_basics.tokenizer import Tokenizer
special_tokens = ['<|endoftext|>']
tinystories_vocab_file = '../vocab_tinystories_sample_5M.txt'
tinystories_merges_file = '../merges_tinystories_sample_5M.txt'
owt_vocab_file = '../owt_after_heap_v2/owt_vocab.json'
owt_merges_file = '../owt_after_heap_v2/owt_merges.txt'

tinystories_tokenizer = Tokenizer.from_files(tinystories_vocab_file,tinystories_merges_file,special_tokens)
owt_tokenizer = Tokenizer.from_files(owt_vocab_file,owt_merges_file,special_tokens)
```

```python
owt_valid_file = '../data/owt_10.txt'
tinystories_valid_file = '../data/TinyStoriesV2-GPT4-10.txt'
import time

def get_ratio(owt_valid_file)->int:
    with open(owt_valid_file,'r') as f_owt:
        owt_contents = f_owt.read().encode('utf-8')
    total_bytes = len(owt_contents)
    print(f"Bytes are {total_bytes}",)

    with open(owt_valid_file,'r') as f_owt:
        encoded = []
        start = time.perf_counter()
        for _id in owt_tokenizer.encode_iterable(f_owt):
            encoded.append(_id)
        end = time.perf_counter()
        encoding_time = (end-start)
        print(f"Encoding time is {encoding_time}",)
        print(f"Bytes per second {total_bytes/encoding_time}")
        print(f"Number of tokens are {len(encoded)}")
        print(f"Ratio is {len(owt_contents)/len(encoded)}")

get_ratio(owt_valid_file)
get_ratio(tinystories_valid_file)
```

```python

def get_ratio_swap(owt_valid_file:str,tokenizer:Tokenizer)->int:
    with open(owt_valid_file,'r') as f_owt:
        owt_contents = f_owt.read().encode('utf-8')
    print(f"Bytes are {len(owt_contents)}",)

    with open(owt_valid_file,'r') as f_owt:
        encoded = []
        for _id in tokenizer.encode_iterable(f_owt):
            encoded.append(_id)
        print(f"Number of tokens are {len(encoded)}")
        print(f"Ratio is {len(owt_contents)/len(encoded)}")

get_ratio_swap(owt_valid_file,tinystories_tokenizer)
get_ratio_swap(tinystories_valid_file,owt_tokenizer)
```

```python
import array
import numpy as np

owt_train_file = '../data/owt_train.txt'
owt_train_outfile='../data/owt_train.npy'
owt_valid_file = '../data/owt_valid.txt'
owt_valid_outfile = '../data/owt_valid.npy'
tinystories_train_file = '../data/TinyStoriesV2-GPT4-train.txt'
tinystories_train_outfile = '../data/TinyStoriesV2-GPT4-train.npy'
tinystories_valid_file = '../data/TinyStoriesV2-GPT4-valid.txt'
tinystories_valid_outfile = '../data/TinyStoriesV2-GPT4-valid.npy'


def serialize_token_ids(infile:str,outfile:str,tokenizer:Tokenizer):
    print(f"Processing {infile} to {outfile}")
    with open(infile,'r') as f:
        stream = (_id for _id in tokenizer.encode_iterable(f))
        # byte_buffer = array.array('H',owt_train_stream).tobytes()
        uint16_array = np.fromiter(stream,dtype=np.uint16)
        np.save(outfile,uint16_array)
        print(f"Numpy array created of size {len(uint16_array)}")

serialize_token_ids(tinystories_valid_file,tinystories_valid_outfile,tinystories_tokenizer)
serialize_token_ids(tinystories_train_file,tinystories_train_outfile,tinystories_tokenizer)
serialize_token_ids(owt_valid_file,owt_valid_outfile,owt_tokenizer)
serialize_token_ids(owt_train_file,owt_train_outfile,owt_tokenizer)

```
